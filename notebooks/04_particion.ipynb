```python
from sklearn.model_selection import train_test_split
```

```python
import pandas as pd
```

```python
df_reid = pd.read_csv('../data/processed/df_reid_resized.csv')
```

```python
ids_unicos = df_reid['dir_name'].unique()
```

```python
train, test = train_test_split(ids_unicos, test_size=0.2, random_state=4)
```

```python
df_train = df_reid[df_reid['dir_name'].isin(train)]
df_test = df_reid[df_reid['dir_name'].isin(test)]
```

```python
df_train.to_csv(
    '../data/processed/train/df_train.csv', index=False
)
```

```python
df_test.to_csv(
    '../data/processed/test/df_test.csv', index=False
)
```
